```python
a = 10
b = 3

print(a + b)   # 13 (加法)
print(a - b)   # 7 (减法)
print(a * b)   # 30 (乘法)
print(a / b)   # 3.3333333333333335 (除法，结果是浮点数)
print(a // b)  # 3 (整除)
print(a % b)   # 1 (余数：10除以3余1)
print(a ** b)  # 1000 (10的3次方)
```

```python
# Python 基本資料型態示範

# 整數 (int)
age = 25
print("age =", age)
print("age 的資料型態是：", type(age))
print()

# 浮點數 (float)
height = 172.5
print("height =", height)
print("height 的資料型態是：", type(height))
print()

# 字串 (str)
name = "Juin-Cheng Lin"
print("name =", name)
print("name 的資料型態是：", type(name))
print()

# 布林值 (bool)
is_student = True
print("is_student =", is_student)
print("is_student 的資料型態是：", type(is_student))
print()

# 額外示範：布林值可用於條件判斷
if is_student:
    print(f"{name} is a student.")
else:
    print(f"{name} is not a student.")
```
