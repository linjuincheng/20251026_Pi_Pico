```python
chinese=int(input("請輸入國文成績："))
print(type(chinese))
english=int(input("請輸入英文成績："))
print(type(english))
math=int(input("請輸入數學成績：")  )
print(type(math))

total=chinese+english+math
print("總分：",total,"分",sep="")
average=round(total/3,ndigits=2)
print("平均：",average,"分",sep="")


```
